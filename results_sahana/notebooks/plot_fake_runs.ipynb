## Trace plots for runs on fake SM database

```python
import numpy as np
import matplotlib.pyplot as plt
import glob
import ast
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import seaborn as sns
import pandas as pd
```

```python
indices = np.arange(1,51)
dictfiles1 = [f"../fake_run/all_dict7.dict"]  #a walk from fake backround (not extreme case)
dictfiles2 = [f"../fake_run/all_dict24.dict"]  #a walk from fake backround (extreme case)
dictfiles = dictfiles1 + dictfiles2 
```

```python
acc_proto = {}
for ind,dfiles in enumerate(dictfiles):
    with open(dfiles,'r') as f:
        di = eval(f.read())
        n_acc = 0
        index = int(dfiles.split('/')[-1].split('t')[1].split('.')[0])
        acc_proto[index] = {'dictlist':[],'n_acc':0,'acc_rate':0.0}
        for i, d in enumerate(di):
            if d == {}:
                print(index)
                continue

            acc = d['Accepted']
            if acc == 0:
                n_acc +=1
                K = d['K']
                TL = d['TL']
                acc_d = {'step': d['step'], 'K':d['K'], 'TL':d['TL']}
                acc_proto[index]['dictlist'].append(acc_d)

    print("\nFile Number ", index)
    print("Number of accepted protomodels: ", n_acc)
    print("Acceptance rate: ", n_acc/di[-1]['step'])
    acc_proto[index]['n_acc'] = n_acc
    acc_proto[index]['acc_rate'] = n_acc/di[-1]['step']
```

```python
def get_ktl_list(dlist):
    K_list,TL_list,step_list = [],[],[]
    for i, d in enumerate(dlist):
        step_list.append(d['step'])
        K_list.append(d['K'])
        TL_list.append(d['TL'])
    
    return K_list, TL_list, step_list

def plot_trace_K(K_list, TL_list, step_list, df):
    fig = make_subplots(rows=2, cols=1, shared_xaxes=True, vertical_spacing=0.07, row_heights=[0.45, 0.55])

    # First subplot (K)
    fig.add_trace(go.Scatter(x=step_list, y=K_list, mode='lines', name='K'), row=1, col=1)

    # Second subplot (TL)
    fig.add_trace(go.Scatter(x=step_list, y=TL_list, mode='lines', name='TL', line=dict(color='red')), row=2, col=1)

    # Update layout
    fig.update_layout(height=500, width=800, showlegend=False, font=dict(family="Computer Modern",size=20),
                        title=dict(text="Evolution of " + "<i>K</i>" + " and " + "<i>TL</i>" + " for walk on a fake SM database",
                        x=0.5,xanchor="center",yanchor="top", font=dict(size=25)))
    
    fig.update_xaxes(title_text="Step", showgrid=True, ticks="outside", row=2, col=1, )
    fig.update_yaxes(title_text=r"<i>K<i>",showgrid=True, ticks="outside", row=1, col=1)
    fig.update_yaxes(title_text=r"<i>TL<i>", showgrid=True, ticks="outside", row=2, col=1)

    

    import plotly.io as pio
    pio.renderers.default = "vscode" 
    pio.show(fig)
    
```

```python
def get_ktl_list(dlist):
    K_list,TL_list,step_list = [],[],[]
    for i, d in enumerate(dlist):
        step_list.append(d['step'])
        K_list.append(d['K'])
        TL_list.append(d['TL'])
    
    return K_list, TL_list, step_list

def plot_trace_K(K_list, TL_list, step_list, df, index):
    fig = make_subplots(rows=2, cols=1, shared_xaxes=True, vertical_spacing=0.07, row_heights=[0.45, 0.55])

    # First subplot (K)
    fig.add_trace(go.Scatter(x=step_list, y=K_list, mode='lines', name='K'), row=1, col=1)

    # Second subplot (TL)
    fig.add_trace(go.Scatter(x=step_list, y=TL_list, mode='lines', name='TL', line=dict(color='red')), row=2, col=1)

    # Update layout
    fig.update_layout(height=500, width=800, showlegend=False, font=dict(family="Computer Modern",size=20),
                        title=dict(text="Evolution of " + "<i>K</i>" + " and " + "<i>TL</i>" + " for walk on a fake SM database",
                        x=0.5,xanchor="center",yanchor="top", font=dict(size=25)))
    if index == 24:
        fig.update_layout(height=500, width=800, showlegend=False, font=dict(family="Computer Modern",size=20),
                        title=dict(text="Evolution of " + "<i>K</i>" + " and " + "<i>TL</i>" + " for walk on a fake SM database (extreme case)",
                        x=0.5,xanchor="center",yanchor="top", font=dict(size=25)))
    fig.update_xaxes(title_text="Step", showgrid=True, ticks="outside", row=2, col=1, )
    fig.update_yaxes(title_text=r"<i>K<i>",showgrid=True, ticks="outside", row=1, col=1)
    fig.update_yaxes(title_text=r"<i>TL<i>", showgrid=True, ticks="outside", row=2, col=1)

    

    import plotly.io as pio
    pio.renderers.default = "vscode" 
    pio.show(fig)
    
```

```python
for ind,proto in acc_proto.items():

    print("Walker ",ind, " % accepted", proto['acc_rate']*100)
    
    df = pd.DataFrame(proto['dictlist'])

    K_list,TL_list,step_list = get_ktl_list(proto['dictlist'])
    print(f"Mean K {round(np.mean(np.array(K_list)),2)}, Mean TL {round(np.mean(np.array(TL_list)),2)}")
    plot_trace_K(K_list, TL_list, step_list,df, ind)
```
